 Exploración y preprocesamiento — Paso 1

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("network_traffic.csv")

print("Dimensiones del dataset:", df.shape)
print("\nTipos de datos:")
print(df.dtypes)
print("\nEstadísticas descriptivas:")
df.describe()
```

Histogramas de bytes_sent y duration_sec

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].hist(df['bytes_sent'], bins=50, color='steelblue', edgecolor='black')
axes[0].set_title('Distribución de bytes_sent')
axes[0].set_xlabel('bytes_sent')
axes[0].set_ylabel('Frecuencia')
axes[0].set_yscale('log')

axes[1].hist(df['duration_sec'], bins=50, color='salmon', edgecolor='black')
axes[1].set_title('Distribución de duration_sec')
axes[1].set_xlabel('duration_sec')
axes[1].set_ylabel('Frecuencia')
axes[1].set_yscale('log')

plt.tight_layout()
plt.savefig("evidencias/SCR-3.1_eda.png", dpi=150)
plt.show()
```

detección y tratamiento de nulos/outliers extremos.

```python
# Verificar valores imposibles (negativos en columnas que no deberían tenerlos)
columnas_numericas = ['bytes_sent', 'bytes_recv', 'duration_sec', 'packets']
for col in columnas_numericas:
    negativos = (df[col] < 0).sum()
    print(f"{col}: {negativos} valores negativos")

# Distribución de la columna label (solo para validación, no se usa en entrenamiento)
print("\nDistribución de 'label':")
print(df['label'].value_counts())
print(df['label'].value_counts(normalize=True) * 100)
```

— Paso 4: Feature Engineering

```python
# Feature engineering: nuevas variables derivadas

# 1. ratio_bytes: proporción entre bytes enviados y recibidos
#    (un ratio muy alto puede indicar exfiltración; muy bajo, descarga masiva)
df['ratio_bytes'] = df['bytes_sent'] / (df['bytes_recv'] + 1)  # +1 evita división por cero

# 2. bytes_por_segundo: tasa de transferencia total (velocidad de la conexión)
df['bytes_por_segundo'] = (df['bytes_sent'] + df['bytes_recv']) / (df['duration_sec'] + 0.01)

# 3. (extra) paquetes_por_segundo: intensidad del tráfico en paquetes
df['paquetes_por_segundo'] = df['packets'] / (df['duration_sec'] + 0.01)

print("Nuevas columnas creadas:")
df[['bytes_sent', 'bytes_recv', 'duration_sec', 'packets',
    'ratio_bytes', 'bytes_por_segundo', 'paquetes_por_segundo']].head(10)
```

: Normalización con StandardScaler

```python
from sklearn.preprocessing import StandardScaler

# Features numéricas a normalizar (excluyendo 'label' que es solo para validación)
features_numericas = ['dst_port', 'bytes_sent', 'bytes_recv', 'duration_sec',
                       'packets', 'ratio_bytes', 'bytes_por_segundo', 'paquetes_por_segundo']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(df[features_numericas])
df_scaled = pd.DataFrame(X_scaled, columns=[f"{c}_scaled" for c in features_numericas])

print("Features normalizadas (primeras filas):")
df_scaled.head()
```

Tarea 3.2 - Entrenamiento del modelo (Isolation Forest)-----------------------------------------------------------------------------


```python
from sklearn.ensemble import IsolationForest

# Entrenamos SOLO con las features normalizadas, excluyendo 'label'
modelo = IsolationForest(contamination=0.05, n_estimators=100, random_state=42)
modelo.fit(X_scaled)

# Predicciones: -1 = anomalía, 1 = normal
df['prediccion'] = modelo.predict(X_scaled)
df['anomaly_score'] = modelo.decision_function(X_scaled)

print("Distribución de predicciones:")
print(df['prediccion'].value_counts())
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score, confusion_matrix, classification_report

# Convertimos label real a formato comparable (-1/1)
y_real = df['label'].map({'anomaly': -1, 'normal': 1})
y_pred = df['prediccion']

precision = precision_score(y_real, y_pred, pos_label=-1)
recall = recall_score(y_real, y_pred, pos_label=-1)
f1 = f1_score(y_real, y_pred, pos_label=-1)

print(f"Precision: {precision:.4f}")
print(f"Recall:    {recall:.4f}")
print(f"F1-Score:  {f1:.4f}")
print("\nReporte de clasificación:")
print(classification_report(y_real, y_pred, target_names=['anomaly', 'normal'], labels=[-1, 1]))
```

```python
cm = confusion_matrix(y_real, y_pred, labels=[-1, 1])

plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['anomaly', 'normal'], yticklabels=['anomaly', 'normal'])
plt.title('Matriz de Confusión - Isolation Forest')
plt.xlabel('Predicción')
plt.ylabel('Real')
plt.tight_layout()
plt.savefig("evidencias/SCR-3.2_metricas.png", dpi=150)
plt.show()
```

Tarea 3.3 — Interpretación y umbral dinámico---------------------------------------------------------------------------

```python
# 1. Score de anomalía para todos los registros
plt.figure(figsize=(12, 5))
plt.plot(df['anomaly_score'].sort_values().values, color='purple')
plt.axhline(0, color='red', linestyle='--', label='Umbral por defecto (0)')
plt.title('Score de anomalía (decision_function) - todos los registros')
plt.xlabel('Registros ordenados por score')
plt.ylabel('Anomaly Score')
plt.legend()
plt.tight_layout()
plt.savefig("evidencias/anomaly_scores.png", dpi=150)
plt.show()
```

```python
# 2. Curva umbral vs F1-Score para encontrar el umbral óptimo
umbrales = np.linspace(df['anomaly_score'].min(), df['anomaly_score'].max(), 100)
f1_scores = []

for u in umbrales:
    pred_u = np.where(df['anomaly_score'] < u, -1, 1)
    f1_u = f1_score(y_real, pred_u, pos_label=-1, zero_division=0)
    f1_scores.append(f1_u)

mejor_idx = np.argmax(f1_scores)
mejor_umbral = umbrales[mejor_idx]
mejor_f1 = f1_scores[mejor_idx]

plt.figure(figsize=(10, 5))
plt.plot(umbrales, f1_scores, color='darkgreen')
plt.axvline(mejor_umbral, color='red', linestyle='--',
            label=f'Umbral óptimo = {mejor_umbral:.4f} (F1={mejor_f1:.4f})')
plt.title('Curva Umbral vs F1-Score')
plt.xlabel('Umbral de anomaly_score')
plt.ylabel('F1-Score')
plt.legend()
plt.tight_layout()
plt.savefig("evidencias/SCR-3.3_umbral_f1.png", dpi=150)
plt.show()

print(f"Umbral óptimo encontrado: {mejor_umbral:.4f}")
print(f"F1-Score con umbral óptimo: {mejor_f1:.4f}")
```

```python
# 3. Top 10 registros más anómalos (score más bajo = más anómalo)
top10_anomalias = df.sort_values('anomaly_score').head(10)
top10_anomalias
```

## Interpretación de las Top 10 anomalías

Los registros con el score de anomalía más bajo (más negativo) representan las
conexiones que el modelo considera más atípicas respecto al comportamiento normal
del tráfico. Estos casos podrían representar una amenaza real porque combinan:

- Volúmenes de `bytes_sent` extremadamente altos en relación al resto del dataset,
  lo cual es consistente con un escenario de exfiltración de datos.
- Duraciones de conexión (`duration_sec`) anormalmente largas, que podrían indicar
  túneles de datos persistentes o conexiones de comando y control (C2).
- Ratios `bytes_sent`/`bytes_recv` desbalanceados, típico de transferencia masiva
  saliente sin una respuesta proporcional del destino.

Estos patrones coinciden con el comportamiento esperado de ataques como
exfiltración de datos, escaneos masivos o tráfico generado por malware.
